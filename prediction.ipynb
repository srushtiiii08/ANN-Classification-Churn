```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
#for prediction we need to load all - the trained model, scaler pickle and onehot pickle

#load the trained model - saved as 'model.h5'
model=load_model('model.h5')

#load the encoder and scaler pickle files
with open('onehot_encoder_geo.pkl', 'rb') as file:
    onehot_encoder_geo_pickle=pickle.load(file)
    
with open('label_encoder_gender.pkl', 'rb') as file:
    label_encoder_gender_pickle=pickle.load(file)
    
with open('scaler.pkl', 'rb') as file:
    scaler_pickle=pickle.load(file)
```

```python
# Example input data  - on which we will do the churn prediction. The csv data was for training & testing
#based on which the model will predict if the new incoming customer will churn(exit) or not
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
#now again, in the input data- we have alphabetical/categorical values that we must convert it to numerical.
#then we have to scale these values based on our scaler; so the data is consistent with the training & testing data

#onehot encode 'Geography'
geo_encoded = onehot_encoder_geo_pickle.transform([[input_data['Geography']]]).toarray()  #gives you a NumPy array with no labels

#Create a DataFrame from geo_encoded, and name its columns using the encoder’s learned feature names.
#first agr=data; second arg: col names to be fetched from fitted pickle file:
#label_encoder_geo is your fitted one‑hot encoder. It knows all categories for the Geography feature from training time.
#get_feature_names_out(['Geography']): means Given an original input feature called Geography, tell me the exact output column names that this encoder produces for that feature
#does not mean “return all columns whose names contain the word Geography.”

#For one‑hot encoding, it turns that one input column (Geography) into three output columns. 
# The encoder internally stores both:which categories exist, and how to name the output columns

geo_encoded_df=pd.DataFrame(geo_encoded, columns=onehot_encoder_geo_pickle.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
#storing the entire input data in dataframe
input_df=pd.DataFrame([input_data])
input_df
```

the Gender column is still a single column named "Gender", but now it contains numbers instead of strings.
No new columns are created, so there is no need to build a new DataFrame or set multiple column names.

Why Geography code is different?
For Geography, you are using a one‑hot encoder → output is a 2D array with multiple columns.
So  reasons:
- used one-hot encoder - bcz we wanted numerical values with multiple columns (not just 0/1)

If we had used label encoder - it would have given as 0,1,2 which is whole another problem, we've discussed previously

```python
#encode categorical values for Gender
input_df['Gender']=label_encoder_gender_pickle.transform(input_df['Gender'])
input_df
```

input_df.drop("Geography", axis=1): axis=1 means “drop a column” (axis 0 would be rows).

pd.concat([A, B], axis=1) means “stick B’s columns to the right of A’s columns, aligning by row index
Here A is input_df without Geography, and B is geo_encoded_df (the one‑hot columns for Geography).

so, Replace the single string column Geography with its one‑hot encoded columns in input_df

```python
#concat one hot encoded geography, while also dropping the og geography column
input_df=pd.concat([input_df.drop('Geography', axis=1), geo_encoded_df], axis=1)
input_df
```

```python
#scaling the input data  -- Scaling is done so that all numeric features are on a similar range, 
# so no single feature dominates just because it has bigger numbers.

input_scaled=scaler_pickle.transform(input_df)
input_scaled
```

```python
#predict chrun
prediction=model.predict(input_scaled)
prediction
```

```python
prediction_prob=prediction[0][0]
prediction_prob
```

```python
if prediction_prob>0.5:
    print("The customer is likely to churn")
else:
    print("The customer is not likely to churn")
```
